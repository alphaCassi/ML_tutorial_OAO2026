# ML Tutorial: full workflow

This notebook checks the GPU, inspects/generated the SPECULA dataset, validates tensor shapes, runs training/testing, and inspects the saved checkpoint. Run it inside the Docker container from the project root.

```python
from pathlib import Path
import sys, subprocess
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name != 'ML_tutorial_OAO2026': PROJECT_ROOT = Path('/workspace/ML_tutorial_OAO2026')
DATA_ROOT = PROJECT_ROOT / 'dataset_specula_ekarus' / 'output' / 'parameter_sweep'
WORKFLOW_ROOT = PROJECT_ROOT / 'ml_workflow'
print(PROJECT_ROOT, DATA_ROOT)
```

## 1. Check Python, PyTorch, CUDA, and CuPy

The simulation uses SPECULA/CuPy on the GPU; training uses PyTorch.

```python
import torch
print('PyTorch:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available(): print('GPU:', torch.cuda.get_device_name(0))
import cupy as cp
print('CuPy:', cp.__version__, 'devices:', cp.cuda.runtime.getDeviceCount())
from cupy.cuda import cublas, cufft
print('CUDA libraries loaded')
```

## 2. Inspect the simulation dataset

Every `sim_*` directory should contain a timestamp directory with `sr_int.fits`, `psfee_int.fits`, `psffwhm_int.fits`, and `params.yml`.

```python
sim_dirs = sorted(DATA_ROOT.glob('sim_*'))
print('Simulation count:', len(sim_dirs))
if sim_dirs:
    sample = next(sim_dirs[0].iterdir())
    print('Example:', sample)
    for f in ['sr_int.fits','psfee_int.fits','psffwhm_int.fits','params.yml']: print(f, (sample/f).exists())
else: print('No simulations found.')
```

## 4. Load and validate one sample

The dense model expects three input features and six target atmospheric parameters.

```python
sys.path.insert(0, str(WORKFLOW_ROOT))
from omegaconf import OmegaConf
from dataset.dataset import AOdataset
cfg = OmegaConf.load(WORKFLOW_ROOT / 'configs' / 'config.yaml')
cfg.dataset.root_data = str(DATA_ROOT)
dataset = AOdataset(cfg)
print('Samples:', len(dataset))
x, y = dataset[0]
print('Input shape:', tuple(x.shape), 'Target shape:', tuple(y.shape))
assert tuple(x.shape) == (3,)
assert tuple(y.shape) == (6,)
```

## 5. Train and test

Hydra uses `configs/mode/train_test.yaml`, so `main.py` trains first and then tests.

```python
subprocess.run([sys.executable, 'main.py'], cwd=WORKFLOW_ROOT, check=True)
```

## 6. Inspect the newest checkpoint

```python
import torch
checkpoints = sorted((WORKFLOW_ROOT/'outputs').glob('**/best_model.pt'), key=lambda p:p.stat().st_mtime)
if checkpoints:
    p = checkpoints[-1]; ckpt = torch.load(p, map_location='cpu'); print(p); print('epoch:', ckpt.get('epoch')); print('best validation loss:', ckpt.get('best_val_loss'))
else: print('No checkpoint found')
```
